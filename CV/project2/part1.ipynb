```python
import numpy as np    # numerical python
import matplotlib     # matrix plotting library
import matplotlib.pyplot # python plotting
import skimage as sk  # sci-kit image
import skimage.io as skio # input/output
import scipy          # scientific python
import cv2            # opencv

import lib
```

```python
imname="data/John.png"
im=skio.imread(imname,as_gray=True)
skio.imshow(im)
skio.show()
lib.save_file("John_original.png",im)
```

```python
d_x=[[1,-1]]
d_y=[[1],[-1]]
pd_x=scipy.signal.convolve2d(im,d_x,mode='same',boundary='symm')
pd_y=scipy.signal.convolve2d(im,d_y,mode='same',boundary='symm')
skio.imshow(pd_x)
skio.show()
lib.save_file('John_pd_x.png',pd_x)
skio.imshow(pd_y)
skio.show()
lib.save_file('John_pd_y.png',pd_y)
```

```python
grad_mag=np.sqrt(np.add(np.square(pd_x),np.square(pd_y)))
grad_mag=np.where(grad_mag>1.0,1.0,grad_mag)
skio.imshow(grad_mag)
skio.show()
lib.save_file('John_grad_mag.png',grad_mag)

for threshold in np.arange(0.1, 1.1, 0.1):
    threshold = round(threshold, 1)  # 避免浮點數精度問題（例如 0.30000000000000004）
    binarized_grad_mag = np.where(grad_mag > threshold, 1.0, 0.0)
    skio.imshow(binarized_grad_mag)
    skio.show()
    lib.save_file(f'John_binarized_grad_mag_{threshold:.1f}.png', binarized_grad_mag)
```

```python
for threshold in np.arange(0.1, 0.2, 0.01):
    threshold = round(threshold, 2)  # 避免浮點數精度問題（例如 0.30000000000000004）
    binarized_grad_mag = np.where(grad_mag > threshold, 1.0, 0.0)
    skio.imshow(binarized_grad_mag)
    skio.show()
    lib.save_file(f'John_binarized_grad_mag_{threshold:.2f}.png', binarized_grad_mag)
```

```python
def make_gaussian(sigma):
    ksize=int(np.ceil(sigma*3)//2*2+1)
    g=cv2.getGaussianKernel(ksize,sigma)
    g_T=np.transpose(g)
    g_filter=np.ma.outer(g,g_T)
    return g_filter
```

```python
sigma=10
g_filter=make_gaussian(sigma)
```

```python
im_blurred=scipy.signal.convolve2d(im,g_filter,mode='full',boundary='wrap')
skio.imshow(im_blurred)
skio.show()
lib.save_file('John_blurred.png',im_blurred)
```

```python
d_x=[[1,-1]]
d_y=[[1],[-1]]

pd_x=scipy.signal.convolve2d(im_blurred,d_x,mode='same',boundary='symm')
skio.imshow(pd_x,cmap='gray')
skio.show()
lib.save_file('John_blurred_pdx.png',pd_x)

pd_y=scipy.signal.convolve2d(im_blurred,d_y,mode='same',boundary='symm')
skio.imshow(pd_y,cmap='gray')
skio.show()
lib.save_file('John_blurred_pdy.png',pd_y)
```

```python
grad_mag=np.sqrt(np.add(np.square(pd_x),np.square(pd_y)))

threshold=0.01
binarized_grad_mag=np.where(grad_mag>threshold,1.0,0.0)
skio.imshow(binarized_grad_mag)
skio.show()
lib.save_file('John_blurred_bi_grad_mad.png',binarized_grad_mag)
```
