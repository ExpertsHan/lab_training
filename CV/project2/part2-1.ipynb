```python
import numpy as np    # numerical python
import matplotlib     # matrix plotting library
import matplotlib.pyplot # python plotting
import skimage as sk  # sci-kit image
import skimage.io as skio # input/output
import scipy          # scientific python
import cv2            # opencv

import lib
```

```python
imname='data/taj.jpg'
im=skio.imread(imname)
b,g,r=cv2.split(im)
skio.imshow(b)
skio.show()
skio.imshow(g)
skio.show()
skio.imshow(r)
skio.show()
```

```python
def make_gaussian(sigma):
    ksize=int(np.ceil(sigma*3)//2*2+1)
    g=cv2.getGaussianKernel(ksize,sigma)
    g_T=np.transpose(g)
    g_filter=np.ma.outer(g,g_T)
    return g_filter
```

```python
sigma=2
g_filter=make_gaussian(sigma)
```

```python
b_blur = scipy.signal.convolve2d(b, g_filter, mode='same', boundary='symm')
g_blur = scipy.signal.convolve2d(g, g_filter, mode='same', boundary='symm')
r_blur = scipy.signal.convolve2d(r, g_filter, mode='same', boundary='symm')

im_blur = cv2.merge([b_blur, g_blur, r_blur])

# 將 0~255 的浮點數轉回 uint8 整數（取代會報錯的 sk.img_as_ubyte）
im_blur = np.clip(im_blur, 0, 255).astype(np.uint8)

skio.imshow(im_blur)  # 注意是 imshow 不是 show
skio.show()
lib.save_file('taj_blur.png',im_blur/255.0)
```

```python
b_high=b-b_blur
g_high=g-g_blur
r_high=r-r_blur
im_high=cv2.merge([b_high,g_high,r_high])
im_high=np.clip(im_high,0,255).astype(np.uint8)
skio.imshow(im_high)
skio.show()
lib.save_file('taj_high.png',im_high/255.0)
```

```python
alpha=3

b_sharp=b+b_high*alpha
g_sharp=g+g_high*alpha
r_sharp=r+r_high*alpha

im_sharp=cv2.merge([b_sharp,g_sharp,r_sharp])
im_sharp=np.clip(im_sharp,0,255).astype(np.uint8)
skio.imshow(im_sharp)
skio.show()
lib.save_file('taj_sharp.png',im_sharp/255.0)
```

```python
imname='data/sharp.png'
im=skio.imread(imname)
b,g,r=cv2.split(im)
b_blur = scipy.signal.convolve2d(b, g_filter, mode='same', boundary='symm')
g_blur = scipy.signal.convolve2d(g, g_filter, mode='same', boundary='symm')
r_blur = scipy.signal.convolve2d(r, g_filter, mode='same', boundary='symm')
b_high=b-b_blur
g_high=g-g_blur
r_high=r-r_blur
alpha=1

b_sharp=b+b_high*alpha
g_sharp=g+g_high*alpha
r_sharp=r+r_high*alpha

im_sharp=cv2.merge([b_sharp,g_sharp,r_sharp])
im_sharp=np.clip(im_sharp,0,255).astype(np.uint8)
skio.imshow(im_sharp)
skio.show()
lib.save_file('sharp_sharp.png',im_sharp/255.0)
```
