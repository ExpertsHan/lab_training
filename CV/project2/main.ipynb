```python
import numpy as np    # numerical python
import matplotlib     # matrix plotting library
import matplotlib.pyplot # python plotting
import skimage as sk  # sci-kit image
import skimage.io as skio # input/output
import scipy          # scientific python
import cv2            # opencv

import lib
```

```python
imname="data/John.png"
im=skio.imread(imname,as_gray=True)
skio.imshow(im)
skio.show()
lib.save_file("John_original.png",im)
```

```python
d_x=[[1,-1]]
d_y=[[1],[-1]]
pd_x=scipy.signal.convolve2d(im,d_x,mode='same',boundary='symm')
pd_y=scipy.signal.convolve2d(im,d_y,mode='same',boundary='symm')
skio.imshow(pd_x)
skio.show()
lib.save_file('John_pd_x.png',pd_x)
skio.imshow(pd_y)
skio.show()
lib.save_file('John_pd_y.png',pd_y)
```

```python
grad_mag=np.sqrt(np.add(np.square(pd_x),np.square(pd_y)))
grad_mag=np.where(grad_mag>1.0,1.0,grad_mag)
skio.imshow(grad_mag)
skio.show()
lib.save_file('John_grad_mag.png',grad_mag)

threshold=0.275
binarized_grad_mag=np.where(grad_mag>threshold,1.0,0.0)
skio.imshow(binarized_grad_mag)
skio.show()
lib.save_file('John_binarized_grad_mag.png',binarized_grad_mag)
```
